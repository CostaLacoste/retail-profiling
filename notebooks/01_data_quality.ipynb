# Практическая работа №1. Первичное профилирование Online Retail II

Для отчёта использованы первые 100 000 строк листа `Year 2010-2011`.

## Пропуски в Customer ID

Идентификатор клиента отсутствует в 34 915 строках из 100 000
(34,915%). Возможные причины — покупки без привязки к профилю,
служебные операции или потеря идентификатора при выгрузке.
Версию о служебных операциях поддерживают примеры строк без
клиента и описания, с нулевой ценой и отрицательным количеством.

## Отрицательные Quantity и Price

Отрицательное количество встречается в 2 079 строках.
У 1 870 из них номер счёта начинается с C. Это согласуется
с гипотезой об отменах или возвратах. Среди остальных есть
записи с нулевой ценой, похожие на складские корректировки.

В выборке отрицательных цен нет. За её пределами на том же листе
найдены счета A563186 и A563187 с ценой −11 062,06 и описанием
Adjust bad debt. Вероятно, это бухгалтерские корректировки
безнадёжной задолженности.
